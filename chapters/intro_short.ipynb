## NLP maps language to useful outputs

| Output | Example task |
|---|---|
| A category or score | sentiment, topic, answerability |
| A label for each unit | part-of-speech tagging, named entities |
| A span or structure | information extraction, parsing, question answering |
| Text in another form | translation, summarisation, dialogue |
| A probability over text | language modelling |

The task determines what counts as a correct output.

## One text can define several tasks

> The battery lasts all day, but the camera is disappointing.

- **Classification:** is the review positive, negative or mixed?
- **Sequence labelling:** which tokens mention product aspects?
- **Information extraction:** battery life → positive; camera → negative
- **Question answering:** what is disappointing? → the camera
- **Generation:** produce a one-sentence summary

The same language does not imply the same supervision or metric.

## A model is one component of an NLP system

1. **Task:** what input produces what output?
2. **Data:** who wrote and annotated the examples, and for which setting?
3. **Representation:** characters, words, subwords or learned vectors?
4. **Model:** rules, linear classifier, neural network or language model?
5. **Evaluation:** which errors matter, and to whom?

A stronger model cannot repair a badly defined task or an invalid evaluation.

## Language variation makes errors systematic

- **Ambiguity:** “I saw her duck” permits different structures and meanings
- **Context:** the intended answer may depend on earlier turns or shared knowledge
- **Domain and register:** clinical notes, social media and news follow different conventions
- **Language and culture:** categories and assumptions do not transfer uniformly
- **Use:** benchmark accuracy may not reflect whether a system helps its users

NLP therefore combines modelling with careful choices about data and evaluation.
